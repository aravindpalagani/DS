```python
course_data = {
    "course_id": ["c101", "c102", "c103", "c104"],
    "course_name": ["Python", "Java", "ML", "DS"],
    "credits": [4, 3, 4, 5],
}
df_courses = pd.DataFrame(course_data)

with pd.ExcelWriter("college_data.xlsx", engine="openpyxl") as writer:
    df_from_excel.to_excel(writer, sheet_name="Students", index=False)
    df_courses.to_excel(writer, sheet_name="courses", index=False)
```
