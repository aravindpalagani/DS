```python
import sqlite3
```

```python
import pandas as pd
```

```python
conn = sqlite3.connect("students.db")
```

```python
cursor = conn.cursor()
```

```python
cursor.execute(
""" 
CREATE TABLE IF NOT EXISTS Students(
    id INTEGER, 
    name TEXT
)
"""
)
```

```python
cursor.execute("INSERT INTO Students VALUES (1, 'John')")
cursor.execute("INSERT INTO Students VALUES (2, 'Reddy')")
cursor.execute("INSERT INTO Students VALUES (3, 'tharun')")
cursor.execute("INSERT INTO Students VALUES (4, 'Kanchana')")

conn.commit()
```

```python
students_df = pd.read_sql_query("SELECT * from Students", conn)
```
